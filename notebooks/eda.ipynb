```python
import matplotlib.pyplot as plt
import numpy as np
import seaborn as sns
```

---

```python
train_pairs_df = pd.read_json('../data/train-grader/my_qa_pairs.json')
train_pairs_df
```

```python
answers = train_pairs_df[train_pairs_df['question'].str.contains('Total Weight')]['answer']
answers = answers.str.split(',', expand=True).astype(float)
# map columns to names
col_map = {0: 'Total Weight'}
answers = answers.rename(columns=col_map)
answers
```

# Plots for Report

```python
# sns.set(font_scale=1.25)
```

```python
plt.rcParams.update({
    'font.size': 14,             # Main font size
    'axes.titlesize': 18,        # Title size
    'axes.labelsize': 16,        # X and Y axis label size
    'xtick.labelsize': 12,       # X tick label size
    'ytick.labelsize': 12,       # Y tick label size
    'legend.fontsize': 12,       # Legend font size
    'figure.titlesize': 20       # Figure title size (for plt.suptitle)
})
```

```python
sns.histplot(data=answers, x='Total Weight', bins=30, kde=True)
plt.title('Distribution of Total Weight in Dishes')
plt.xlabel('Total Weight (g)')
plt.ylabel('Number of Dishes')
ymax = 3000
plt.ylim(0, ymax)
plt.tight_layout()
# save figure
plt.savefig('../report/imgs/total_weight_dist.png')
plt.show()
```

```python
sns.histplot(data=train_df, x='Iron, Fe', bins=30, kde=True)
plt.title('Distribution of Iron Content in Dishes')
plt.xlabel('Iron, Fe (mg)')
plt.ylabel('Number of Dishes')
ymax = 3000
plt.ylim(0, ymax)
plt.tight_layout()
# save figure
plt.savefig('../report/imgs/iron_dist.png')
plt.show()
```

```python
sns.histplot(data=train_df, x='Calcium, Ca', bins=30, kde=True)
plt.title('Distribution of Calcium Content in Dishes')
plt.xlabel('Calcium, Ca (mg)')
plt.ylabel('Number of Dishes')
ymax = 3000
plt.ylim(0, ymax)
plt.tight_layout()
# save figure
plt.savefig('../report/imgs/calcium_dist.png')
plt.show()
```

```python
sns.histplot(data=train_df, x='Vitamin C, total ascorbic acid', bins=30, kde=True)
plt.title('Distribution of Vitamin C Content in Dishes')
plt.xlabel('Vitamin C, total ascorbic acid (mg)')
plt.ylabel('Number of Dishes')
ymax = 3000
plt.ylim(0, ymax)
plt.tight_layout()
# save figure
plt.savefig('../report/imgs/vitC_dist.png')
plt.show()
```

```python
# binsize = int(np.sqrt(len(train_df)))
binsize = 30

fig, axs = plt.subplots(2, 2, figsize=(15, 5))
sns.histplot(data=answers, x='Total Weight', kde=True, bins=binsize, ax=axs[0, 0])
axs[0, 0].set_title('Distribution of Total Weight in Dishes')
axs[0, 0].set_xlabel('Total Weight (g)')
axs[0, 0].set_ylabel('Number of Dishes')
sns.histplot(data=train_df, x='Iron, Fe', kde=True, bins=binsize, ax=axs[0, 1])
axs[0, 1].set_title('Distribution of Iron Content in Dishes')
axs[0, 1].set_xlabel('Iron, Fe (mg)')
axs[0, 1].set_ylabel('Number of Dishes')
sns.histplot(data=train_df, x='Calcium, Ca', kde=True, bins=binsize, ax=axs[1, 0])
axs[1, 0].set_title('Distribution of Calcium Content in Dishes')
axs[1, 0].set_xlabel('Calcium, Ca (mg)')
axs[1, 0].set_ylabel('Number of Dishes')
sns.histplot(data=train_df, x='Vitamin C, total ascorbic acid', kde=True, bins=binsize, ax=axs[1, 1])
axs[1, 1].set_title('Distribution of Vitamin C Content in Dishes')
axs[1, 1].set_xlabel('Vitamin C, total ascorbic acid (mg)')
axs[1, 1].set_ylabel('Number of Dishes')
# share the same y-axis scale
ymax = 3000
axs[0, 0].set_ylim(0, ymax)
axs[0, 1].set_ylim(0, ymax)
axs[1, 0].set_ylim(0, ymax)
axs[1, 1].set_ylim(0, ymax)
plt.tight_layout()
plt.show()
```

```python
# binsize = int(np.sqrt(len(train_df)))
binsize = 30

# vertical histplots for Total Weight, Iron, Calcium, Vitamin C
fig, axs = plt.subplots(4, 1, figsize=(6, 10))

sns.histplot(data=answers, x='Total Weight', kde=True, bins=binsize, ax=axs[0])
axs[0].set_title('Distribution of Total Weight in Dishes')
axs[0].set_xlabel('Total Weight (g)')
axs[0].set_ylabel('Number of Dishes')

sns.histplot(data=train_df, x='Iron, Fe', kde=True, bins=binsize, ax=axs[1])
axs[1].set_title('Distribution of Iron Content in Dishes')
axs[1].set_xlabel('Iron, Fe (mg)')
axs[1].set_ylabel('Number of Dishes')

sns.histplot(data=train_df, x='Calcium, Ca', kde=True, bins=binsize, ax=axs[2])
axs[2].set_title('Distribution of Calcium Content in Dishes')
axs[2].set_xlabel('Calcium, Ca (mg)')
axs[2].set_ylabel('Number of Dishes')

sns.histplot(data=train_df, x='Vitamin C, total ascorbic acid', kde=True, bins=binsize, ax=axs[3])
axs[3].set_title('Distribution of Vitamin C Content in Dishes')
axs[3].set_xlabel('Vitamin C, total ascorbic acid (mg)')
axs[3].set_ylabel('Number of Dishes')

# share the same y-axis scale
ymax = 3000
axs[0].set_ylim(0, ymax)
axs[1].set_ylim(0, ymax)
axs[2].set_ylim(0, ymax)
axs[3].set_ylim(0, ymax)

plt.tight_layout()
plt.show()
```
